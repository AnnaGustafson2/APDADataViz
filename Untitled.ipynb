# Participation in American Parliamentary Debate Association (APDA)

```python
import duckdb
import pandas as pd

df_rounds = pd.read_csv("Data_Curation_Data/Rounds_Final_Debater_Restoration.csv")
df_rounds = df_rounds.rename(columns={"Speaker One School": "Speaker_One_School", "W/L": "W_L", "Judge Years Experience": "Judge_Years_Experience"})
con = duckdb.connect()
con.register('rounds', df_rounds)

query = """
SELECT Speaker_One_School
FROM rounds
GROUP BY Speaker_One_School
ORDER BY Speaker_One_School ASC
"""
schools_df = con.execute(query).df()
print(schools_df.head())
```
