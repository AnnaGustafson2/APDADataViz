# Analysis of American Parlimentary Debate Association Rounds Fall 2019 - Spring 2026

## Exploratory Data Analysis -- Judging Burden

While collecting the round data for this project, I noticed that there is a subset of judges who are neither registered with a school nor have they competed in any round themselves. This means a number of judges are not registered and do not compete in ADPA rounds. The following section will outline this occurence and analyze aspects of judging accountability.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import display, HTML
from scipy import stats
```

```python
df_rounds = pd.read_csv("Rounds.csv")
df_rounds_with_judge_id = pd.read_csv("Rounds_With_Identified_Judges.csv")
debate_roster = pd.read_csv("All_Debaters.csv")
# BROKEN, DELETED FILE
unidentified_judges = pd.read_csv("Lost_Rounds.csv")
```

```python
df_rounds_with_judge_id["Judge Identified"] = True
unidentified_judges["Judge Identified"] = False
df_rounds_data = pd.concat([df_rounds_with_judge_id, unidentified_judges], ignore_index=True)

fig, ax = plt.subplots(figsize=(10, 5), layout="constrained")

sns.countplot(
    data=df_rounds_data,
    x="Judge Identified",
    orient="v",
    color="mediumslateblue",
    order=df_rounds_data["Judge Identified"].value_counts().index,
    ax=ax,
)
ax.set_title("Rounds with Judge Associated with a School")
ax.set_ylabel("Number of Rounds")

plt.show()
```

```python
num_ided = len(df_rounds_data[df_rounds_data["Judge Identified"] == True])
num_unided = len(df_rounds_data[df_rounds_data["Judge Identified"] == False])
print("Number of Identified Judges: ", num_ided, "\nNumber of Unidentified Judges: ", num_unided, "\nPercentage Unidentified: ", num_unided / (num_ided + num_unided))
```

From the above we can see that 13.49% of all rounds are conducted with a judge who never have competed and are not registered with any school. This might seem like a low percentage overall, however, since unfair or unwarrented decisions are one of the biggest reasons why novices feel discouraged from returning to more tournaments, having judges with a minimum amount of experience will only work to improve retention.

In the following sections, I will analyze where this 13.49% is coming from, and how we might improve the judging pool moving forward.

```python
unidentified_judges["Semester"] = unidentified_judges["Season"] + " " + unidentified_judges["Year"].astype(str)
ordered_semesters = ["Fall 2019", "Spring 2020", "Fall 2020", "Spring 2021", "Fall 2021", "Spring 2022", "Fall 2022", "Spring 2023",
                     "Fall 2023", "Spring 2024", "Fall 2024", "Spring 2025", "Fall 2025", "Spring 2026"]

fig, ax = plt.subplots(figsize=(10, 5), layout="constrained")

sns.countplot(
    data=unidentified_judges,
    x="Semester",
    order=ordered_semesters,
    orient="v",
    color="palevioletred",
    ax=ax,
)
ax.set_title("Rounds With Unidentified Judge By Semester")
ax.set_ylabel("Number of Unidentified Judge Rounds")
ax.tick_params(axis="x", rotation=30)

plt.show()
```

```python
counts = unidentified_judges["Semester"].value_counts().reindex(ordered_semesters).rename("Unidentified Judge")
print(counts.to_frame(), "\nMean number of rounds with unidentified judges: ", counts.mean())
```

From the above, we can see that an average of 155 rounds per semester are conducted with an unregistered and noncompeting judge. We can also see a significant spike to 372 rounds with unidentified judges at Fall 2019. This could be explained by a two compounding factors that will be described below

(1) Systemic Entry Issues

While collecting and processing data for this analysis, I encountered an unexpected systemic underreporting issue in the debater rosters within the APDA website. Each debater who competes is expected to be registered under their respective school on the APDA website; however, scraping each school's roster revealed many active debaters who were not registered under any school at all. In one case, a debater managed to compete in 17 different tournaments without registration. To correct this, I manually cross-referenced each unregistered debater against their partners' listed schools and appended entries to the scraped roster, inferring school affiliation from partnership records where direct registration was missing. In all, I appended 989 debaters who competed between Fall 2019 and Spring 2026. 

Additionally, fearing the quality of records surrounding "Year on the Team" the APDA website's tracker of years the debater has competed, I recalculated values of "Initial Year Seen" and "Final Year Seen" by searching for actual occurances of each name and replacing the roster's entry if my results produced a larger range.

```python
# Roster before my appending
debate_roster_before_recovery = pd.read_csv("Debater_Roster.csv")
active_before_recovery = debate_roster_before_recovery[debate_roster_before_recovery["Final Year Seen"] >= 2019]

# Roster after my recovery sorted
active_in_frame = debate_roster[debate_roster["Final Year Seen"] >= 2019]

num_roster_pre = len(active_before_recovery)
num_roster_post = len(active_in_frame)
print("Number of debaters on the APDA roster from Fall 2019 to Spring 2026: ", num_roster_pre, "\nNumber of debaters on the updated roster from Fall 2019 to Spring 2026: ", num_roster_post,
     "\nNumber of debaters gained: ", num_roster_post - num_roster_pre, "\nPercentage of debaters added over total: ", (num_roster_post - num_roster_pre) / num_roster_post)
```

The result is that 1092 debaters who previously were not on the roster between Fall 2019 and Spring 2026 were now on the roster for this timeframe due to my algorithms. This is 29.98% of all the debaters seen between Fall 2019 and Spring 2026. This indicate a serious lack of proper documentation that needs to be corrected moving forward.  

Additionally, this could explain the jump in missing debaters in Fall 2019. Since my methods relied on analyzing previous rounds, many debaters who competed mostly before Fall 2019 (data I do not have access to) and were not registered, would not have been reconstructed by my analysis.

(2) COVID-19 

To further reduce the likelihood of unregistered debaters occurring in Fall 2019 from being recovered by my algorithms, the effects of COVID-19 begining in Fall 2019 may have exacerbated this issue. Debaters from Fall 2019 - Spring 2021 may have been discouraging debaters from participating in tournaments due to safety concerns, school closures, or disconnection from their schools, not to mention the general reduction in avaliable tournaments during this period. Without sufficient records of tournaments attended and partner entries, I can not make entries for unidentified judges.

In order to check the state of unidentified judges without the possible interferrence from COVID-19 or edge of the dataset, I will look at the counts again, but only for Fall 2021 - Spring 2026

```python
unidentified_judges["Semester"] = unidentified_judges["Season"] + " " + unidentified_judges["Year"].astype(str)
ordered_semesters = ["Fall 2021", "Spring 2022", "Fall 2022", "Spring 2023", "Fall 2023", "Spring 2024", "Fall 2024", 
                     "Spring 2025", "Fall 2025", "Spring 2026"]

fig, ax = plt.subplots(figsize=(10, 5), layout="constrained")

sns.countplot(
    data=unidentified_judges,
    x="Semester",
    order=ordered_semesters,
    orient="v",
    color="palevioletred",
    ax=ax,
)
ax.set_title("Rounds With Unidentified Judge By Semester")
ax.set_ylabel("Number of Unidentified Judge Rounds")
ax.tick_params(axis="x", rotation=30)

plt.show()
```

```python
counts = unidentified_judges["Semester"].value_counts().reindex(ordered_semesters).rename("Unidentified Judge")
print(counts.to_frame(), "\nMean number of rounds with unidentified judges: ", counts.mean())
```

```python

# WORKING HERE!

unidentified_judges["Season_only"] = unidentified_judges["Semester"].str.split().str[0]
unidentified_judges["Year_only"] = unidentified_judges["Semester"].str.split().str[1].astype(int)

fall = counts[counts.index.str.startswith("Fall")]
spring = counts[counts.index.str.startswith("Spring")]
stat, p = stats.ttest_rel(fall.values, spring.values[:len(fall)])  # paired if same # of years
print(stat, p)
```
