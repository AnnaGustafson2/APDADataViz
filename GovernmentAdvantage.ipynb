```python
import duckdb
import pandas as pd

df_rounds = pd.read_csv("Data_Curation_Data/Rounds_Final_Debater_Restoration.csv")
df_rounds = df_rounds.rename(columns={"G/O": "G_O", "W/L": "W_L", "Judge Years Experience": "Judge_Years_Experience"})
con = duckdb.connect()
con.register('rounds', df_rounds)

query = """
SELECT COUNT(*) as Wins, Judge_Years_Experience
FROM rounds
WHERE (
    ((W_L == 'W') AND (G_O == 'G')) OR
    ((W_L == 'L') AND (G_O == 'O'))
)
GROUP BY Judge_Years_Experience
ORDER BY Judge_Years_Experience ASC
"""
gov_wins = con.execute(query).df()

query = """
SELECT COUNT(*) as Wins, Judge_Years_Experience
FROM rounds
WHERE (
    ((W_L == 'L') AND (G_O == 'G')) OR
    ((W_L == 'W') AND (G_O == 'O'))
)
GROUP BY Judge_Years_Experience
ORDER BY Judge_Years_Experience ASC
"""
opp_wins = con.execute(query).df()
```

```python
wins = pd.merge(gov_wins, opp_wins, on="Judge_Years_Experience", how="inner")
wins = wins.rename(columns={"Wins_x": "Gov_Wins", "Wins_y": "Opp_Wins"})
wins["Gov_Overal"] = wins["Gov_Wins"] / (wins["Gov_Wins"] + wins["Opp_Wins"])
print(wins.head(8))
```
